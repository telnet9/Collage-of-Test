```python
import tkinter as tk

class QuickSnakeFill:
    def __init__(self, n=10, size=40):
        self.n, self.size = n, size
        self.root = tk.Tk()
        self.root.title("蛇形填数")
    
        self.canvas = tk.Canvas(self.root, width=n*size, height=n*size, bg="white")
        self.canvas.pack(padx=10, pady=10)

        self.matrix = [[0] * n for _ in range(n)]
        self.dirs = [(0, 1), (1, 0), (0, -1), (-1, 0)] 
        self.x, self.y, self.d, self.curr_num = 0, 0, 0, 1

        self.step()
        self.root.mainloop()

    def step(self):
        if self.curr_num > self.n * self.n: return 

        x1, y1 = self.y * self.size, self.x * self.size
        self.canvas.create_rectangle(x1, y1, x1+self.size, y1+self.size, fill="#ADD8E6", outline="gray")
        self.canvas.create_text(x1 + self.size/2, y1 + self.size/2, text=str(self.curr_num), font=("Arial", 12, "bold"))

        self.matrix[self.x][self.y] = self.curr_num
        self.curr_num += 1
        

        dx, dy = self.dirs[self.d]
        nx, ny = self.x + dx, self.y + dy
        if not (0 <= nx < self.n and 0 <= ny < self.n and self.matrix[nx][ny] == 0):
            self.d = (self.d + 1) % 4
            dx, dy = self.dirs[self.d]
            nx, ny = self.x + dx, self.y + dy
            
        self.x, self.y = nx, ny
        self.root.after(100, self.step)

if __name__ == "__main__":
    QuickSnakeFill(10)
```
