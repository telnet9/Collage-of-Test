```python
#lanqiao 分巧克力
def check(x):
    count = 0
    for val in arr:
        min_l = min(val[0], val[1])
        if x > min_l:
            continue
        count += (val[0] // x) * (val[1] // x)
    if count >= k:
        return True
    else:
        return False

n, k = map(int, input().split())
arr = []
for i in range(n):
    arr.append(list(map(int, input().split())))
left, right = 1, 100000
while left + 1 < right:
    mid = (left + right) // 2
    if check(mid):
        left = mid
    else:
        right = mid
if check(right):
    print(right)
else:
    print(left)
```

```python
#实验报告第三题
import pandas as pd
import numpy as np
s9=pd.Series([65, 92, 48, np.nan, 86, 73, 105, np.nan])
print(s9)
print(s9.isnull())
has_missing = s9.isnull().any()
print("数据中是否存在缺失值：", has_missing)
total = s9.sum()
avg = s9.mean()
print("总和：", total)
print("平均值：", avg)
above_80 = s9[s9 > 80]  
print(above_80)
count_80 = len(above_80) 
print("大于80的个数：", count_80)
```

```python
#判断可分解数
def main():
  n=int(input())
  nums=map(int,input().split())
  cnt=0
  for x in nums:
    if x == 1:
      cnt += 1
  print(n-cnt)
if __name__=="__main__":
  main()
```
