```python
import pandas as pd
data = {    
    "姓名": ["张海洋", "孟晓舟", "毛不易", "司浩然", "潘特", "胡丹", "杜威伟", "苏晓", "胡丹", "路啸"],
    "年龄": [68, 34, 16, 18, 44, 52, 19, 32, 52, 22],
    "体重": ["70公斤", "154斤", "76公斤", "78公斤", "198斤", "189斤", "56公斤", "78公斤", "189斤", "45公斤"],
    "性别": ["m", "f", "f", "m", "f", "f", "f", "m", "女", "m"],
    "00~06": [72, 85, 65, 78, 69, 68, 71, 78, 68, 92],
    "06~12": [69, 84, 69, 79, None, 75, 78,None, 75, 95],
    "12~18": [71, 76, 72, 72, 75, 72, 75, 75, 131, 87]
}
df = pd.DataFrame(data)
df
```

```python
#2、查看有哪些列有缺失值。
```

```python
print("每列缺失值数量：")
print(df.isnull().sum())

print("\n哪些列存在缺失值：")
print(df.isnull().any())
```

```python
#3、缺失值可以用平均值替代。
```

```python
df["06~12"].mean()
```

```python
df["06~12"] = df["06~12"].fillna(df["06~12"].mean())
df
```

```python
#统一数据类型，把心率统一为整数型。
```

```python
df["06~12"] = df["06~12"].round().astype(int)
```

```python
df
```

```python
# 统一性别：把“男”→m，“女”→f
df['性别'] = df['性别'].replace({'男': 'm', '女': 'f'})

# 查看修改后的性别列
print("统一后的性别列：")
print(df['性别'])
```

```python
df
```

```python
# 1. 提取纯数字
num_weight = df['体重'].str.extract(r'(\d+)').astype(float)

# 2. 空列表
new_weights = []

# 3. 循环 + if 判断单位
for i in range(len(df)):
    w = df['体重'].iloc[i]
    
   
    num = num_weight.iloc[i,0]
    
    if "公斤" in w:
        kg = num
    else:
        kg = num*0.5
    
    new_weights.append(int(kg))

# 4. 覆盖
df['体重'] = new_weights
```

```python
df
```

```python
check_cols = ["00~06", "06~12", "12~18"]

for col in check_cols:
    mean_val = df[col].mean()
    
    # 找出异常
    abnormal = df[(df[col] > mean_val + 30) | (df[col] < mean_val - 30)]
    
    print(f"列：{col}  平均值：{mean_val:.1f}")
    if not abnormal.empty:
        print("异常数据：")
        print(abnormal[["姓名", col]])
    else:
        print("无异常")
    print("-"*40)
```

```python
check_cols = ["00~06", "06~12", "12~18"]

for col in check_cols:
    mean_val = df[col].mean()
    mask = (df[col] > mean_val + 30) | (df[col] < mean_val - 30)
    df.loc[mask, col] = int(round(mean_val))

df
```
