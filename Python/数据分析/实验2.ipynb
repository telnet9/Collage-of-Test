```python
import pandas as pd

data = {
    "客户编号": [1, 1, 2, 2, 3, 4, 4, 5],
    "商家": ["云海肴", "丰收日", "云海肴", "丰收日", "西秦美食", "西秦美食", "丰收日", "云海肴"],
    "品名": ["过桥米线", "小炒肉", "过桥米线", "红烧肉", "羊肉汤", "肉夹馍", "牛肉粒", "汽锅鸡"],
    "数量": [2, 1, 1, 1, 1, 2, 1, 1],
    "单价": [49, 38, 49, 68, 33, 18, 48, 88]
}

df1 = pd.DataFrame(data)
df1
```

```python
import pandas as pd

# 构造数据
data = {
    "客户编号": [1, 2, 3, 4, 5],
    "姓名": ["张洋", "段小海", "李景鹏", "李亮", "王霞"],
    "性别": ["男", "女", "男", "男", "女"],
    "地址": ["绥化学院9公寓401", "绥化学院4公寓201", "绥化学院9公寓501", "绥化学院9公寓301", "绥化学院4公寓301"]
}

df2 = pd.DataFrame(data)
df2
```

```python
user_order = pd.merge(df1, df2, on="客户编号")
user_order
```

```python
result = user_order[user_order["品名"] == "过桥米线"]["姓名"]
result
```

```python
result1 = user_order[(user_order["商家"] == "云海肴") | (user_order["商家"] == "丰收日")]["姓名"].drop_duplicates() 
result1
```

```python
user_order["小计"] = user_order["数量"] * user_order["单价"]
user_order
```

```python
per_customer_total = user_order.groupby(["客户编号","姓名"])["小计"].sum()
print("每位客户订餐总金额：")
print(per_customer_total)

# 2. 所有订单总金额
all_total = user_order["小计"].sum()
print(f"\n所有订单总金额：{all_total}")
```

```python
shop_result = user_order.groupby("商家").agg(
    订单数量=("小计", "count"),    # 统计订单笔数
    平均金额=("小计", "mean")     # 统计平均金额
)

print("每个商家的订单数量和平均金额：")
print(shop_result)
```

```python
shops = user_order["商家"].unique()
print("订单来自的商家：")
print(shops)
```

```python
result_df = user_order[["姓名", "品名", "数量", "小计"]]

# 2. 按 小计 降序排序（金额从大到小）
result_df = result_df.sort_values(by="小计", ascending=False)

# 输出结果
print("按小计降序排序后：")
print(result_df)
```
